```python

def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def get_neighbors(state):

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        ("RIGHT", 0, 1),
        ("DOWN", 1, 0),
        ("LEFT", 0, -1),
        ("UP", -1, 0)
    ]

    for move, dr, dc in moves:

        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:

            new_zero = nr * 3 + nc
            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            yield move, tuple(new_state)


def depth_limited_dfs(state, goal, limit, path):

    # Goal reached
    if state == goal:
        return True

    # Depth limit reached
    if limit == 0:
        return False

    for move, next_state in get_neighbors(state):

        # Avoid cycles in current path
        if next_state not in [s for m, s in path]:

            path.append((move, next_state))

            if depth_limited_dfs(
                next_state,
                goal,
                limit - 1,
                path
            ):
                return True

            # Backtrack
            path.pop()

    return False


def IDS(start, goal, max_depth):

    for limit in range(max_depth + 1):

        path = [("START", start)]

        if depth_limited_dfs(
            start,
            goal,
            limit,
            path
        ):
            return path

    return None


# -------------------------
# MAIN
# -------------------------

start = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

goal = (
   1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

path = IDS(start, goal, 5)


if path:

    for move, state in path:

        if move == "START":
            print("START")
        else:
            print("↓", move)

        print_state(state)

    print("GOAL REACHED")

else:
    print("NO SOLUTION")
print("submitted by 1BM24CS110")
```
